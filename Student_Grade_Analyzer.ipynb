```python
# Function to calculate grade and pass/fail status
def evaluate_performance(marks):
    total = sum(marks)
    average = total / len(marks)

    if average >= 90:
        grade = "A+"
    elif average >= 75:
        grade = "A"
    elif average >= 60:
        grade = "B"
    elif average >= 40:
        grade = "C"
    else:
        grade = "F"

    # Pass/Fail check (agar average >= 40 aur har subject mein >= 35 ho)
    if average >= 40 and all(m >= 35 for m in marks):
        status = "PASS"
    else:
        status = "FAIL"

    return total, average, grade, status

# Function to take student input
def get_student_data():
    student_data = {}
    while True:
        name = input("Enter Student Name76: ").strip()
        if name.lower() == 'done':
            break

        num_subjects = int(input(f"Enter number of subjects for {name}: "))
        marks = []
        for i in range(num_subjects):
            score = float(input(f"  Enter score for subject {i + 1}: "))
            marks.append(score)

        student_data[name] = marks
    return student_data

# Function to display formatted report
def display_report(student_data):
    print("\n" + "=" * 68)
    print(f"{'STUDENT ACADEMIC PERFORMANCE REPORT':^68}")
    print("=" * 68)
    print(f"{'Student Name':<20} {'Total':<10} {'Average':<10} {'Grade':<10} {'Status':<10}")
    print("-" * 68)

    for name, marks in student_data.items():
        total, average, grade, status = evaluate_performance(marks)
        print(f"{name:<20} {total:<10.2f} {average:<10.2f} {grade:<10} {status:<10}")

    print("=" * 68)

# Main execution
data = get_student_data()
if data:
    display_report(data)
else:
    print("No student data entered.")
```
